## **Questão 1:**
<font size=3>

Resolva a tarefa de **regressão** do *dataset* [`fetch_california_housing`](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.fetch_california_housing.html) com o modelo [`LinearRegression`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html). Siga o **fluxo de trabalho** de aprendizado de máquina, descrito na aula.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt


from sklearn.datasets import fetch_california_housing # Dados para o modelo

from sklearn.model_selection import train_test_split #para treino e test
from sklearn.linear_model import LinearRegression #o modelo propriamente dito

from sklearn.preprocessing import StandardScaler
```

```python


california = fetch_california_housing() #objeto para os dados
X = california.data # dados do objeto
Y = california.target # alvo destino

df = pd.DataFrame(X, columns=california.feature_names) # dataset para fazer a exploraxao dos dados
df["MedHouseVal"] = Y

x_train, x_test, y_train, y_test = train_test_split(X,Y , test_size=0.3, train_size=0.7) # Divide os dados em dados de test e dados de train

print(f"X-train:{x_train.shape}, y-train:{y_train.shape}") # printa esses dados
print(f"X-test:{x_test.shape}, y-test:{y_test.shape}")


scale = StandardScaler() # para fazer a normalizaçao dos dados
model = LinearRegression() # modelo escolhido

x_train_scale = scale.fit_transform(x_train) # Normaliza os dados e fit
x_test_scale = scale.transform(x_test) # apenas normaliza dos dados de test

model.fit(x_train_scale,y_train) # treina o modelo


y_pred_test = model.predict(x_test_scale) # previsao apartir dos dados de test

plt.scatter(x_test[:,0], y_test, color="tab:orange", alpha=0.5, label="Real (Teste)")
plt.scatter(x_test[:,0], y_pred_test, color="tab:blue", alpha=0.5, label="Previsto (Teste)") # plota o grafico

plt.xlabel("Renda Médio (Escalonada)")
plt.ylabel("Preço da Casa")
plt.legend()



plt.show()


plt.scatter(y_test, y_pred_test, alpha=0.3)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel("Preço Real")
plt.ylabel("Preço Previsto")

plt.show()

df.head()
```

## **Questão 2:**
<font size=3>

Resolva a tarefa de **classificação** do *dataset* [`load_iris`](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_iris.html) com o modelo [`LogisticRegression`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html). Siga o **fluxo de trabalho** de aprendizado de máquina, descrito na aula.

```python
from sklearn.datasets import load_iris # Dados para o modelo
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from matplotlib.colors import ListedColormap
```

```python
# 1. Carregar e usar apenas as duas primeiras colunas para o gráfico 2D
iris = load_iris()
X = iris.data[:, :2]  # <--- Selecionamos apenas 2 colunas aqui
y = iris.target

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

model = LogisticRegression(max_iter=2000)
model.fit(x_train, y_train)

# 2. Criar o grid para o fundo
x_min, x_max = x_train[:, 0].min() - 1, x_train[:, 0].max() + 1
y_min, y_max = x_train[:, 1].min() - 1, x_train[:, 1].max() + 1
atri1, atri2 = np.meshgrid(np.arange(x_min, x_max, 0.02),
                           np.arange(y_min, y_max, 0.02))

# Agora o predict recebe 2 colunas, condizente com o treino
z = model.predict(np.c_[atri1.ravel(), atri2.ravel()])
z = z.reshape(atri1.shape)

# 3. Plotagem
cmap_fundo = ListedColormap(["#FFCCCC", "#CCCCFF", "#CCFFCC"]) # 3 cores para 3 classes
cmap_pontos = ListedColormap(["red", "blue", "green"])

plt.figure(figsize=(8, 6))
plt.contourf(atri1, atri2, z, alpha=0.3, cmap=cmap_fundo)

plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train, cmap=cmap_pontos,
            edgecolor="k", s=30, label="Treino")
plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test, cmap=cmap_pontos,
            marker="^", edgecolor="k", s=50, label="Teste")

plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title("Fronteiras de Decisão - Regressão Logística (Iris)")
plt.legend()
plt.show()

y_pred = model.predict(x_test)
acc = accuracy_score(y_test, y_pred)

print(f"Acurácia: {acc:.2f}")
```
